```python
dbutils.widgets.dropdown(
    "environment",
    "DEV",
    ["DEV", "STAGING", "PROD"]
)

dbutils.widgets.text(
    "delivery_date",
    ""
)

dbutils.widgets.text(
    "delivery_batch",
    ""
)

dbutils.widgets.dropdown(
    "force_reprocess",
    "false",
    ["false", "true"]
)
```

```python
# =========================================================
# 1. LEER PARÁMETROS
# =========================================================

environment = (
    dbutils.widgets.get(
        "environment"
    )
    .strip()
    .upper()
)

delivery_date = (
    dbutils.widgets.get(
        "delivery_date"
    )
    .strip()
)

delivery_batch = (
    dbutils.widgets.get(
        "delivery_batch"
    )
    .strip()
)

force_reprocess = (
    dbutils.widgets.get(
        "force_reprocess"
    )
    .strip()
    .lower()
    == "true"
)


# =========================================================
# 2. CONFIGURAR ENVIRONMENT
# =========================================================

configure_environment(
    environment
)

require_environment()


# =========================================================
# 3. VALIDAR PARÁMETROS DEL BATCH
# =========================================================

if not delivery_date:

    raise ValueError(
        "delivery_date es obligatorio"
    )


if not delivery_batch:

    raise ValueError(
        "delivery_batch es obligatorio"
    )


if delivery_batch not in [
    "0800",
    "1600",
    "2300"
]:

    raise ValueError(
        f"INVALID_DELIVERY_BATCH | "
        f"delivery_batch={delivery_batch}"
    )


# =========================================================
# 4. CONSTRUIR BATCH ID
# =========================================================

batch_id = (
    delivery_date.replace(
        "-",
        ""
    )
    + "_"
    + delivery_batch
)


# =========================================================
# 5. MOSTRAR CONTEXTO
# =========================================================

print(
    f"environment={ENVIRONMENT} | "
    f"catalog={CATALOG} | "
    f"delivery_date={delivery_date} | "
    f"delivery_batch={delivery_batch} | "
    f"batch_id={batch_id} | "
    f"force_reprocess={force_reprocess}"
)
```

```python
process_silver_batch(
    batch_id=batch_id,
    force_reprocess=force_reprocess
)
```
